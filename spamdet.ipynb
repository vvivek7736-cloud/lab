```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
df=pd.read_csv("email.csv")
df
```

```python
df.groupby('spam')
df.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.text,df.spam,test_size=0.20)
```

```python
x_train
```

```python
y_train
```

```python
v=CountVectorizer()
x_train_count=v.fit_transform(x_train.values)
x_train_count.toarray()[:3]
```

```python
model=MultinomialNB()
model.fit(x_train_count,y_train)
```

```python
email=['hey vivek can we go to watch a football match tommorow','do not have money,exclusive offer ust for you , dont miss this reward!!']
email_count=v.transform(email)
```

```python
model.predict(email_count)
```

```python
x_test_count=v.transform(x_test)
model.score(x_test_count,y_test)
```
